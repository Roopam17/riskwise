```python
# Cell 1: a table of three invented daily prices, using pandas ("Excel in code")
import pandas as pd

prices = pd.DataFrame({
    "day": ["Mon", "Tue", "Wed"],
    "price": [100, 103, 101],
})
prices
```

```python
# Cell 2: how much did the price change each day, in percent?
# (This "daily return" is the building block of volatility.)
prices["change_pct"] = prices["price"].pct_change() * 100
prices
```

```python
# Cell 3: draw the prices as a line chart
import matplotlib.pyplot as plt

plt.plot(prices["day"], prices["price"], marker="o")
plt.title("Three invented daily prices")
plt.ylabel("price")
plt.show()
```
